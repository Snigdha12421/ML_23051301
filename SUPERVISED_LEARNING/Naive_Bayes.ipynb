```python
X = [
    ["Sunny", "Hot"],
    ["Sunny", "Hot"],
    ["Overcast", "Hot"],
    ["Rain", "Mild"],
    ["Rain", "Cool"],
    ["Rain", "Cool"],
    ["Overcast", "Cool"],
    ["Sunny", "Mild"]
]

y = ["No", "No", "Yes", "Yes", "Yes", "No", "Yes", "No"]
```

```python
def train_naive_bayes(X, y):
    model = {}
    class_counts = {}
    total_samples = len(y)

    for label in y:
        class_counts[label] = class_counts.get(label, 0) + 1

    for c in class_counts:
        model[c] = {
            "prior": class_counts[c] / total_samples,
            "features": {}
        }

    for i in range(len(X[0])): 
        for c in class_counts:
            model[c]["features"][i] = {}

        for row, label in zip(X, y):
            feature_value = row[i]
            model[label]["features"][i][feature_value] = \
                model[label]["features"][i].get(feature_value, 0) + 1

        for c in class_counts:
            total = class_counts[c]
            for value in model[c]["features"][i]:
                model[c]["features"][i][value] /= total

    return model
```

```python
def predict(model, sample):
    probabilities = {}

    for c in model:
        prob = model[c]["prior"]
        for i, value in enumerate(sample):
            prob *= model[c]["features"][i].get(value, 1e-6)
        probabilities[c] = prob

    return max(probabilities, key=probabilities.get)
```

```python
model = train_naive_bayes(X, y)

test_sample = ["Sunny", "Cool"]
result = predict(model, test_sample)

print("Prediction:", result)
```
