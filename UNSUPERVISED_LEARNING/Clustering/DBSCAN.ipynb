```python
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.metrics.pairwise import euclidean_distances

# Dataset
X = np.array([
    [1,2],
    [2,2],
    [2,3],
    [8,7],
    [8,8],
    [25,80]
])

eps = 2
min_samples = 2

dist_matrix = euclidean_distances(X, X)

neighbors = []
for i in range(len(X)):
    neigh = []
    for j in range(len(X)):
        if dist_matrix[i][j] <= eps:
            neigh.append(X[j])
    neighbors.append(neigh)

db = DBSCAN(eps=eps, min_samples=min_samples)
db.fit(X)

labels = db.labels_
core_indices = db.core_sample_indices_

core_points = X[core_indices]
noise_points = X[labels == -1]

border_points = []
for i in range(len(X)):
    if i not in core_indices and labels[i] != -1:
        border_points.append(X[i])

print("Neighbors of each point:")
for i in range(len(X)):
    print(X[i], "->", neighbors[i])

print("\nCore Points:")
print(core_points)

print("\nBorder Points:")
print(border_points)

print("\nNoise Points:")
print(noise_points)
```
