```python
import numpy as np

X = np.array([
    [1,2],
    [1,4],
    [1,0],
    [10,2],
    [10,4],
    [10,0]
])

def distance(a,b):
    return np.sqrt(np.sum((a-b)**2))

def split_cluster(cluster):

    max_dist = 0
    p1, p2 = None, None

    for i in range(len(cluster)):
        for j in range(i+1,len(cluster)):
            d = distance(cluster[i],cluster[j])
            if d > max_dist:
                max_dist = d
                p1, p2 = cluster[i], cluster[j]

    c1 = []
    c2 = []

    for point in cluster:
        if distance(point,p1) < distance(point,p2):
            c1.append(point)
        else:
            c2.append(point)

    return np.array(c1), np.array(c2)

clusters = [X]

k = 2

while len(clusters) < k:
    cluster_to_split = clusters.pop(0)
    c1,c2 = split_cluster(cluster_to_split)
    clusters.append(c1)
    clusters.append(c2)

# print clusters
for i,cluster in enumerate(clusters):
    print("Cluster",i+1)
    print(cluster)
```
